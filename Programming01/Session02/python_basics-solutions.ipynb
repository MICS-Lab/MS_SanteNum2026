## 1. Conditions: `if`, `elif`, `else`

Classify a systolic blood pressure `sbp` into `"low"` ($< 90$), `"normal"` ($< 120$), `"elevated"` ($< 130$) or `"high"` (otherwise), with `if` / `elif` / `else`.
Test it with `sbp = 85`, `118`, `125` and `150`.

````python
for sbp in [85, 118, 125, 150]:
    if sbp < 90:
        category = "low"
    elif sbp < 120:
        category = "normal"
    elif sbp < 130:
        category = "elevated"
    else:
        category = "high"
    print(sbp, category)
````

````python
temp = 39.0
"fever" if temp >= 38 else "no fever"
````

## 3. `while` loops, `break` & `continue`

A drug has a half-life of $6$ hours: every $6$ hours, its concentration is divided by $2$.
Starting from $100$ mg/L, how many hours until the concentration falls below $5$ mg/L? Use a `while` loop.

````python
concentration = 100
hours = 0
while concentration >= 5:
    concentration /= 2
    hours += 6
hours, concentration
````

Loop over `glucose = [5.1, 5.8, 6.3, 12.4, 5.5, 4.9]` (mmol/L) and stop at the first value above $11$ (`break`). Print its position.

````python
glucose = [5.1, 5.8, 6.3, 12.4, 5.5, 4.9]
for i, g in enumerate(glucose):
    if g > 11:
        print(f"hyperglycemia at position {i}: {g}")
        break
````

Some measurements are missing: `values = [72, None, 85, None, 64]`. Compute their mean, skipping the missing values with `continue`.

````python
values = [72, None, 85, None, 64]
total, count = 0, 0
for v in values:
    if v is None:
        continue
    total += v
    count += 1
total / count
````

Write a function `mean(values, skip_none=True)` with a default argument: when `skip_none` is `True`, missing values are ignored.
Call it on `values` with and without the default.

````python
def mean(values, skip_none=True):
    if skip_none:
        values = [v for v in values if v is not None]
    return sum(values) / len(values)


mean(values)
````

Rewrite the class with:
- a method `bmi()` returning the BMI of the patient,
- a method `__repr__` returning a readable string, like `Patient('P001', 54 y)`.

````python
class Patient:
    def __init__(self, id, age, weight, height):
        self.id = id
        self.age = age
        self.weight = weight
        self.height = height

    def bmi(self):
        return self.weight / self.height**2

    def __repr__(self):
        return f"Patient({self.id!r}, {self.age} y)"


p = Patient("P001", 54, 81.5, 1.78)
p, p.bmi()
````

Add to the class:
- an attribute `heart_rates`, an empty list created in `__init__`,
- a method `add_heart_rate(value)` that appends a measurement,
- a method `mean_heart_rate()` that returns the mean, or `None` when there is no measurement.

````python
class Patient:
    def __init__(self, id, age, weight, height):
        self.id = id
        self.age = age
        self.weight = weight
        self.height = height
        self.heart_rates = []

    def bmi(self):
        return self.weight / self.height**2

    def add_heart_rate(self, value):
        self.heart_rates.append(value)

    def mean_heart_rate(self):
        if len(self.heart_rates) == 0:
            return None
        return sum(self.heart_rates) / len(self.heart_rates)

    def __repr__(self):
        return f"Patient({self.id!r}, {self.age} y)"


p = Patient("P001", 54, 81.5, 1.78)
print(p.mean_heart_rate())
for h in [72, 85, 64]:
    p.add_heart_rate(h)
p.mean_heart_rate()
````

## 6. Mini use case

Write a class `Ward` with:
- a `name` and a `capacity` (maximum number of patients),
- a list of `patients`, empty at creation,
- a method `admit(patient)` that adds the patient if the ward is not full, and prints a message otherwise,
- a method `__len__` returning the number of patients (so that `len(ward)` works).

````python
class Ward:
    def __init__(self, name, capacity):
        self.name = name
        self.capacity = capacity
        self.patients = []

    def admit(self, patient):
        if len(self.patients) >= self.capacity:
            print(f"{self.name} is full: cannot admit {patient.id}")
        else:
            self.patients.append(patient)

    def __len__(self):
        return len(self.patients)
````

Run the cell below, which creates a ward of capacity $3$ and tries to admit $4$ patients with some heart rate measurements.

````python
ward = Ward("Cardiology", capacity=3)
data = [
    ("P001", 54, 81.5, 1.78, [72, 85, 64]),
    ("P002", 37, 62.0, 1.65, [110, 118, 105]),
    ("P003", 71, 95.2, 1.70, []),
    ("P004", 29, 55.4, 1.60, [58, 61]),
]
for id, age, weight, height, rates in data:
    patient = Patient(id, age, weight, height)
    for h in rates:
        patient.add_heart_rate(h)
    ward.admit(patient)
len(ward), ward.patients
````

Add a method `alerts()` to `Ward` that returns the list of ids of patients with tachycardia (mean heart rate $> 100$) or with no measurement at all.

````python
class Ward:
    def __init__(self, name, capacity):
        self.name = name
        self.capacity = capacity
        self.patients = []

    def admit(self, patient):
        if len(self.patients) >= self.capacity:
            print(f"{self.name} is full: cannot admit {patient.id}")
        else:
            self.patients.append(patient)

    def __len__(self):
        return len(self.patients)

    def alerts(self):
        result = []
        for patient in self.patients:
            mean_hr = patient.mean_heart_rate()
            if mean_hr is None or mean_hr > 100:
                result.append(patient.id)
        return result


ward = Ward("Cardiology", capacity=5)
data = [
    ("P001", 54, 81.5, 1.78, [72, 85, 64]),
    ("P002", 37, 62.0, 1.65, [110, 118, 105]),
    ("P003", 71, 95.2, 1.70, []),
    ("P004", 29, 55.4, 1.60, [58, 61]),
]
for id, age, weight, height, rates in data:
    patient = Patient(id, age, weight, height)
    for h in rates:
        patient.add_heart_rate(h)
    ward.admit(patient)

ward.alerts()
````

Add a method `report()` that prints one line per patient, like `P001 | 54 y | BMI 25.7 (overweight) | HR 73.7`, using the function `bmi_category` from section 4. Write `HR n/a` when there is no measurement.

````python
def report(self):
    print(f"Ward {self.name} ({len(self)}/{self.capacity})")
    for patient in self.patients:
        value = patient.bmi()
        mean_hr = patient.mean_heart_rate()
        hr = "n/a" if mean_hr is None else f"{mean_hr:.1f}"
        print(
            f"{patient.id} | {patient.age} y | BMI {value:.1f} ({bmi_category(value)}) | HR {hr}"
        )


Ward.report = report
ward.report()
````
