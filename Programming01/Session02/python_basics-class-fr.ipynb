## 3. `while` boucles, `break` & `continue`

Un médicament a une demi-vie de $6$ heures : toutes les $6$ heures, sa concentration est divisée par $2$.
À partir de $100$ mg/L, combien d'heures jusqu'à ce que la concentration tombe en dessous de $5$ mg/L ? Utilisez une boucle `while`.

````python
c = 100
t = 0
while c > 5:
    c /= 2
    t += 6
print(f"t={t}, c={c}")
````

Écrivez une fonction `mean(values, skip_none=True)` avec un argument par défaut : lorsque `skip_none` vaut `True`, les valeurs manquantes sont ignorées.
Appelez-le sur `values` avec et sans la valeur par défaut.

````python
def mean(vals, skip_none=True):
    tot = 0
    c = 0
    for v in vals:
        if skip_none:
            if v is not None:
                tot += v
                c += 1
        else:
            if v is None:
                v = 0
            tot += v
            c += 1
    return tot / c


mean([72, None, 85, None, 64], skip_none=False)
# mean([72, 0, 85, 0, 64]) # equivalent
````

Réécrivez la classe avec :
- une méthode `bmi()` renvoyant l'IMC du patient,
- une méthode `__repr__` renvoyant une chaîne lisible, comme `Patient('P001', 54 y)`.

````python
class Patient:
    def __init__(self, id, age, weight, height):
        self.id = id
        self.age = age
        self.weight = weight
        self.height = height

    def bmi(self):
        return self.weight / self.height**2

    def __repr__(self):
        return f"Patient {self.id}"


p1 = Patient("P001", 54, 81.5, 1.78)
p2 = Patient("P002", 64, 71.5, 1.72)
print(p1, p2)
print(p1.bmi())
````

Ajouter à la classe :
- un attribut `heart_rates`, une liste vide créée en `__init__`,
- une méthode `add_heart_rate(value)` qui annexe une mesure,
- une méthode `mean_heart_rate()` qui renvoie la moyenne, ou `None` lorsqu'il n'y a pas de mesure.

````python
class Patient:
    def __init__(self, id, age, weight, height, heart_rates=[]):
        self.id = id
        self.age = age
        self.weight = weight
        self.height = height
        self.heart_rates = heart_rates

    def add_heart_rate(self, v):
        self.heart_rates.append(v)

    def mean_heart_rates(self):
        return mean(self.heart_rates)

    def bmi(self):
        return self.weight / self.height**2

    def __repr__(self):
        return f"Patient {self.id}"


p1 = Patient("P001", 54, 81.5, 1.78)
print(p1.heart_rates)
p1.add_heart_rate(50)
p1.add_heart_rate(60)
print(p1.heart_rates)
print(p1.mean_heart_rates())
````

Écrivez une classe `InPatient` qui **hérite** de `Patient` et stocke également un nombre `room`.
Son `__init__` devrait appeler celui de `Patient` (`super().__init__`). Vérifiez qu'un `InPatient` a toujours une méthode `bmi()` et testez `isinstance`.

````python
class InPatient(Patient):
    def __init__(self, id, age, weight, height, room):
        super().__init__(id, age, weight, height)
        self.room = room

    def __repr__(self):
        return f"Inpatient({self.id!r}, {self.age} y, room {self.room})"


ip = InPatient("P003", 71, 95.2, 1.70, room=12)
ip, ip.bmi(), isinstance(ip, Patient), isinstance(p1, InPatient)
````
