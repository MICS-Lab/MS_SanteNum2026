## 3. `while` loops, `break` & `continue`

A drug has a half-life of $6$ hours: every $6$ hours, its concentration is divided by $2$.
Starting from $100$ mg/L, how many hours until the concentration falls below $5$ mg/L? Use a `while` loop.

````python
c = 100
t = 0
while c > 5:
    c /= 2
    t += 6
print(f"t={t}, c={c}")
````

````python
def bmi_category(val):
    if val < 18.5:
        return "underweight"
    if val < 25:
        return "normal"
    if val < 30:
        return "overweight"
    else:
        return "obese"


bmi_category(22)
````

Write a function `mean(values, skip_none=True)` with a default argument: when `skip_none` is `True`, missing values are ignored.
Call it on `values` with and without the default.

````python
def mean(vals, skip_none=True):
    tot = 0
    c = 0
    for v in vals:
        if skip_none:
            if v is not None:
                tot += v
                c += 1
        else:
            if v is None:
                v = 0
            tot += v
            c += 1
    return tot / c


mean([72, None, 85, None, 64], skip_none=False)
# mean([72, 0, 85, 0, 64]) # equivalent
````

Rewrite the class with:
- a method `bmi()` returning the BMI of the patient,
- a method `__repr__` returning a readable string, like `Patient('P001', 54 y)`.

````python
class Patient:
    def __init__(self, id, age, weight, height):
        self.id = id
        self.age = age
        self.weight = weight
        self.height = height

    def bmi(self):
        return self.weight / self.height**2

    def __repr__(self):
        return f"Patient {self.id}"


p1 = Patient("P001", 54, 81.5, 1.78)
p2 = Patient("P002", 64, 71.5, 1.72)
print(p1, p2)
print(p1.bmi())
````

Add to the class:
- an attribute `heart_rates`, an empty list created in `__init__`,
- a method `add_heart_rate(value)` that appends a measurement,
- a method `mean_heart_rate()` that returns the mean, or `None` when there is no measurement.

````python
class Patient:
    def __init__(self, id, age, weight, height, heart_rates=[]):
        self.id = id
        self.age = age
        self.weight = weight
        self.height = height
        self.heart_rates = heart_rates

    def add_heart_rate(self, v):
        self.heart_rates.append(v)

    def mean_heart_rates(self):
        return mean(self.heart_rates)

    def bmi(self):
        return self.weight / self.height**2

    def __repr__(self):
        return f"Patient {self.id}"


p1 = Patient("P001", 54, 81.5, 1.78)
print(p1.heart_rates)
p1.add_heart_rate(50)
p1.add_heart_rate(60)
print(p1.heart_rates)
print(p1.mean_heart_rates())
````

````python
class InPatient(Patient):
    def __init__(self, id, age, weight, height, room):
        super().__init__(id, age, weight, height)
        self.room = room

    def __repr__(self):
        return f"Inpatient({self.id!r}, {self.age} y, room {self.room})"


ip = InPatient("P003", 71, 95.2, 1.70, room=12)
ip, ip.bmi(), isinstance(ip, Patient), isinstance(p1, InPatient)
````
