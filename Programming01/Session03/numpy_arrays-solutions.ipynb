## 8. Why vectorize?

Compute the sum of the squares of the integers from $0$ to $10^6 - 1$, first with a Python `for` loop, then with NumPy.
Compare the timings with `%timeit`.

```python
def sum_squares_loop(n):
    total = 0
    for i in range(n):
        total += i * i
    return total


def sum_squares_numpy(n):
    x = np.arange(n)
    return (x * x).sum()


n = 1_000_000
sum_squares_loop(n), sum_squares_numpy(n)
```

```python
%timeit sum_squares_loop(n)
%timeit sum_squares_numpy(n)
```

## 9. Mini use case

We simulate a cohort of $200$ patients. Run the cell below, which creates `height` (in meters) and `weight` (in kilograms).

```python
rng = np.random.default_rng(42)
n_patients = 200
height = rng.normal(1.70, 0.09, n_patients)
weight = rng.normal(72, 14, n_patients)
```

Compute the body mass index of every patient, $\mathrm{BMI} = \mathrm{weight} / \mathrm{height}^2$, then its mean, minimum and maximum.

```python
bmi = weight / height**2
bmi.mean(), bmi.min(), bmi.max()
```

What fraction of patients is obese ($\mathrm{BMI} \geq 30$)? What fraction is in the normal range ($18.5 \leq \mathrm{BMI} < 25$)?

*Hint: the mean of a boolean array is a fraction.*

```python
(bmi >= 30).mean(), ((bmi >= 18.5) & (bmi < 25)).mean()
```

Assign each patient a category: $0$ = underweight ($< 18.5$), $1$ = normal, $2$ = overweight ($\geq 25$), $3$ = obese ($\geq 30$), using `np.digitize`.
Count the patients in each category (`np.bincount`).

```python
category = np.digitize(bmi, [18.5, 25, 30])
np.bincount(category, minlength=4)
```

Plot the histogram of the BMI (`plt.hist`), with vertical lines at $18.5$, $25$ and $30$ (`plt.axvline`).

```python
plt.hist(bmi, bins=30)
for threshold in [18.5, 25, 30]:
    plt.axvline(threshold, color="k", linestyle="--")
plt.xlabel("BMI (kg/m²)")
plt.ylabel("Number of patients")
plt.show()
```
