Slicing and assignment in numpy.

Create a $4 \times 4$ array of ones, then set its column $1$ to $0$ and its row $2$ to $-1$.
Print the first row, the first column and the last row.

```python
x = np.ones( (4, 4) )
x[:, 1:3] = 0
x[2, :] = -1
x[0,0] = 3
x[:, 0], x[-1, :]
```

Concatenate three copies of that array, first along `axis=1`, then along `axis=0` (`np.concat`).
What is the shape of each result?

```python
x, np.concat([x, x, x], axis=0)
```
