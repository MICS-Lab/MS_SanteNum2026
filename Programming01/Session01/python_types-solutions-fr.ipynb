## 1. Les nombres : `int` et `float`

Imprimez le type de `42`, `42.0` et `4.2e1` (`type`). `42` et `42.0` sont-ils égaux (`==`) ?

```python
type(42), type(42.0), type(4.2e1), 42 == 42.0
```

Calculez `7 / 2`, `7 // 2`, `7 % 2` et `7 ** 2`. Quel est le type de chaque résultat ?

```python
7 / 2, 7 // 2, 7 % 2, 7 ** 2  # / always returns a float, the others keep int
```

```python
type(7 / 2), type(7 // 2), type(7 % 2), type(7 ** 2)
```

Calculez $2^{100}$. Y a-t-il une taille maximale pour un Python `int` ? Calculez maintenant `2.0 ** 1024` : que se passe-t-il avec un `float` ?

```python
2 ** 100  # ints have arbitrary precision
```

```python
# 2.0 ** 1024  # OverflowError: floats are 64 bits, max about 1.8e308
```

Calculez `0.1 + 0.2`. Est-ce égal à `0.3` ? Comment comparer deux floats ?

*Indice : `math.isclose` (vous pouvez `import math` ici).*

```python
0.1 + 0.2, 0.1 + 0.2 == 0.3
```

```python
import math

math.isclose(0.1 + 0.2, 0.3), abs((0.1 + 0.2) - 0.3) < 1e-9
```

```python
round(3.14159, 2), int(3.9), int(-3.9)  # int truncates towards zero
```

## 9. Mini cas d'utilisation

Exécutez la cellule ci-dessous, qui crée une liste de dossiers de patients (un dictionnaire par patient).

```python
patients = [
    {"id": "P001", "age": 54, "weight": 81.5, "height": 1.78, "diagnoses": ["hypertension", "diabetes"]},
    {"id": "P002", "age": 37, "weight": 62.0, "height": 1.65, "diagnoses": []},
    {"id": "P003", "age": 71, "weight": 95.2, "height": 1.70, "diagnoses": ["diabetes"]},
    {"id": "P004", "age": 29, "weight": 55.4, "height": 1.60, "diagnoses": ["asthma"]},
    {"id": "P005", "age": 63, "weight": None, "height": 1.82, "diagnoses": ["hypertension"]},
]
```

Calculez l'âge moyen des patients.

```python
ages = [p["age"] for p in patients]
sum(ages) / len(ages)
```

Construisez la liste des identifiants des patients âgés de plus de $50$.

```python
[p["id"] for p in patients if p["age"] > 50]
```

Créez un dictionnaire mappant chaque identifiant de patient à son IMC, arrondi à une décimale. Évitez les patients dont le poids est manquant.

```python
{p["id"]: round(p["weight"] / p["height"] ** 2, 1) for p in patients if p["weight"] is not None}
```

Comptez combien de patients ont chaque diagnostic, dans un dictionnaire `{diagnosis: count}`.

*Astuce : parcourez les patients et leurs diagnostics, et utilisez `counts.get(d, 0) + 1`.*

```python
counts = {}
for p in patients:
    for d in p["diagnoses"]:
        counts[d] = counts.get(d, 0) + 1
counts
```

Imprimez une ligne par patient, au format `P001 | 54 y | hypertension, diabetes` (écrivez `none` lorsqu'il n'y a pas de diagnostic).

*Indice : `", ".join(...)`.*

```python
for p in patients:
    diagnoses = ", ".join(p["diagnoses"]) or "none"
    print(f"{p['id']} | {p['age']} y | {diagnoses}")
```
