## 9. Mini use case

Run the cell below, which creates a list of patient records (one dictionary per patient).

```python
patients = [
    {"id": "P001", "age": 54, "weight": 81.5, "height": 1.78, "diagnoses": ["hypertension", "diabetes"]},
    {"id": "P002", "age": 37, "weight": 62.0, "height": 1.65, "diagnoses": []},
    {"id": "P003", "age": 71, "weight": 95.2, "height": 1.70, "diagnoses": ["diabetes"]},
    {"id": "P004", "age": 29, "weight": 55.4, "height": 1.60, "diagnoses": ["asthma"]},
    {"id": "P005", "age": 63, "weight": None, "height": 1.82, "diagnoses": ["hypertension"]},
]
```

Compute the mean age of the patients.

```python
ages = [p["age"] for p in patients]
sum(ages) / len(ages)
```

Build the list of ids of the patients older than $50$.

```python
[p["id"] for p in patients if p["age"] > 50]
```

Build a dictionary mapping each patient id to its BMI, rounded to one decimal. Skip patients whose weight is missing.

```python
{p["id"]: round(p["weight"] / p["height"] ** 2, 1) for p in patients if p["weight"] is not None}
```

Count how many patients have each diagnosis, in a dictionary `{diagnosis: count}`.

*Hint: loop over the patients and their diagnoses, and use `counts.get(d, 0) + 1`.*

```python
counts = {}
for p in patients:
    for d in p["diagnoses"]:
        counts[d] = counts.get(d, 0) + 1
counts
```

Print one line per patient, formatted like `P001 | 54 y | hypertension, diabetes` (write `none` when there is no diagnosis).

*Hint: `", ".join(...)`.*

```python
for p in patients:
    diagnoses = ", ".join(p["diagnoses"]) or "none"
    print(f"{p['id']} | {p['age']} y | {diagnoses}")
```
